# *args and **kwargs

```python
# *args
# allows us to pass a variable number of non-keyword argument to a function.
```

```python
def multiply(*args):  # you can use any name at the args but always use with *
    product =1;

    for i in args:
        product *= i

    return product
```

```python
multiply(3,4,5,6,2,3,7,5)  # you can give multiple inputs
```

```python
# **kwargs
# **kwargs allows us to pass any number of keyword arguments
# keyword argument mean that contain a key-value pair, like a Python dictionary
```

```python
def demo(**kwargs):
    print(kwargs)

demo(Name="Rakesh",Age=18,course="BCA")
```

```python
def display(**kwargs):
    for (key,value) in kwargs.items():
        print(key,"->",value)
```

```python
display(Name="Rakesh",Age=18,course="BCA")
```

Point to remember while using *args and **kwargs

    . oder of the argument matter(normal-> *args -> **kwargs)
    .The words " args" and "kwargs" are only a convention, you can use any name of your choice

# Lambda Function

```python
# x -> x^2

a = lambda x:x**2
a(3)
```

```python
# x,y -> x+y

b = lambda x,y:x+y
b(3,4)
```

```python
# check if a string has 'a'

a = lambda s:'a' in s
a('hello')
```

```python
# odd or even

a = lambda x:'even' if x%2 == 0 else 'odd'
a(10)
```

# Higher Order Function

```python
# Example

def square(x):
    return x**2

# HOF
def transform(f,L):
    output = []

    for i in L:
        output.append(f(i))

    print(output)

L = [1,2,3,4,5]

transform(square,L)
```

```python
transform (lambda x:x**3,L)
```

# Map

```python
# square the items of a list

map(lambda x:x**2,[1,2,3,4,5])
```

```python
list(map(lambda x:x**2,[1,2,3,4,5]))
```

```python
# odd/even labelling of list items
L = [1,2,3,4,5]  
list(map(lambda x:"even" if x%2 == 0 else "odd",L))
```

# Filter

```python
# Numbers greater than 5

L = [3,4,5,6,7]

list(filter(lambda x:x>5,L))
```

```python
# fetch fruits strating with 'a'

fruits = ['apple','mango','guava']

list(filter(lambda x:x.startswith('a'),fruits))
```

# Reduce

```python
import functools

# sum of all items

functools.reduce(lambda x,y:x+y,[1,2,3,4,5])
```

```python
# find min

functools.reduce(lambda x,y:x if x<y else y,[3,2,4,3,5,6])
```
